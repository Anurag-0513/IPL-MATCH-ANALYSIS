```python


import numpy as np
import pandas as pd

# IPL MATCH ANALYSIS PROJECT (2024-2026)
# Made using pandas and numpy
# Concepts: DataFrames, NumPy arrays, functions, while loop, if-elif-else
# No graph or visualization is used.

# -------------------- DATA --------------------

# Each season row:
# [year, winner, runner-up, Orange Cap player, runs,
#  Purple Cap player, wickets, fours, sixes, total boundaries]

season_data = [
    ["2024", "Kolkata Knight Riders", "Sunrisers Hyderabad",
     "Virat Kohli (RCB)", 741, "Harshal Patel (PBKS)", 24, 2174, 1260, 3434],

    ["2025", "Royal Challengers Bengaluru", "Punjab Kings",
     "Sai Sudharsan (GT)", 759, "Prasidh Krishna (GT)", 25, 2245, 1294, 3539],

    ["2026", "Royal Challengers Bengaluru", "Gujarat Titans",
     "Vaibhav Sooryavanshi (RR)", 776, "Kagiso Rabada (GT)", 29, 2332, 1426, 3758]
]

season_df = pd.DataFrame(season_data, columns=[
    "Year", "Winner", "Runner-up", "Orange Cap", "Orange Cap Runs",
    "Purple Cap", "Purple Cap Wickets", "Fours", "Sixes", "Total Boundaries"
])


# Each playoff row:
# [year, match, scorecard, venue, winner, result, player of the match]

playoff_data = [
    ["2024", "Qualifier 1", "SRH 159; KKR 164/2", "Narendra Modi Stadium, Ahmedabad", "Kolkata Knight Riders", "KKR won by 8 wickets", "Mitchell Starc (KKR)"],
    ["2024", "Eliminator", "RCB 172/8; RR 174/6", "Narendra Modi Stadium, Ahmedabad", "Rajasthan Royals", "RR won by 4 wickets", "Ravichandran Ashwin (RR)"],
    ["2024", "Qualifier 2", "SRH 175/9; RR 139/7", "M. A. Chidambaram Stadium, Chennai", "Sunrisers Hyderabad", "SRH won by 36 runs", "Shahbaz Ahmed (SRH)"],
    ["2024", "Final", "SRH 113; KKR 114/2", "M. A. Chidambaram Stadium, Chennai", "Kolkata Knight Riders", "KKR won by 8 wickets", "Mitchell Starc (KKR)"],

    ["2025", "Qualifier 1", "PBKS 101; RCB 106/2", "Maharaja Yadavindra Singh Stadium, Mullanpur", "Royal Challengers Bengaluru", "RCB won by 8 wickets", "Josh Hazlewood (RCB)"],
    ["2025", "Eliminator", "MI 228/5; GT 208/6", "Maharaja Yadavindra Singh Stadium, Mullanpur", "Mumbai Indians", "MI won by 20 runs", "Rohit Sharma (MI)"],
    ["2025", "Qualifier 2", "MI 203/6; PBKS 207/5", "Narendra Modi Stadium, Ahmedabad", "Punjab Kings", "PBKS won by 5 wickets", "Shreyas Iyer (PBKS)"],
    ["2025", "Final", "RCB 190/9; PBKS 184/7", "Narendra Modi Stadium, Ahmedabad", "Royal Challengers Bengaluru", "RCB won by 6 runs", "Krunal Pandya (RCB)"],

    ["2026", "Qualifier 1", "RCB 254/5; GT 162", "HPCA Stadium, Dharamshala", "Royal Challengers Bengaluru", "RCB won by 92 runs", "Rajat Patidar (RCB)"],
    ["2026", "Eliminator", "RR 243/8; SRH 196", "Maharaja Yadavindra Singh Stadium, New Chandigarh", "Rajasthan Royals", "RR won by 47 runs", "Vaibhav Sooryavanshi (RR)"],
    ["2026", "Qualifier 2", "RR 214/6; GT 219/3", "Maharaja Yadavindra Singh Stadium, New Chandigarh", "Gujarat Titans", "GT won by 7 wickets", "Shubman Gill (GT)"],
    ["2026", "Final", "GT 155/8; RCB 161/5", "Narendra Modi Stadium, Ahmedabad", "Royal Challengers Bengaluru", "RCB won by 5 wickets", "Virat Kohli (RCB)"]
]

playoff_df = pd.DataFrame(playoff_data, columns=[
    "Year", "Match", "Scorecard", "Venue", "Winner", "Result", "Player of the Match"
])


# Each player row: [year, category, rank, player, runs/wickets/sixes]

player_data = [
    # IPL 2024 - Run scorers
    ["2024", "Runs", 1, "Virat Kohli (RCB)", 741],
    ["2024", "Runs", 2, "Ruturaj Gaikwad (CSK)", 583],
    ["2024", "Runs", 3, "Riyan Parag (RR)", 573],
    ["2024", "Runs", 4, "Travis Head (SRH)", 567],
    ["2024", "Runs", 5, "Sanju Samson (RR)", 531],

    # IPL 2024 - Wicket takers
    ["2024", "Wickets", 1, "Harshal Patel (PBKS)", 24],
    ["2024", "Wickets", 2, "Varun Chakaravarthy (KKR)", 21],
    ["2024", "Wickets", 3, "Jasprit Bumrah (MI)", 20],
    ["2024", "Wickets", 4, "T Natarajan (SRH)", 19],
    ["2024", "Wickets", 5, "Harshit Rana (KKR)", 19],

    # IPL 2024 - Six hitters
    ["2024", "Sixes", 1, "Abhishek Sharma (SRH)", 42],
    ["2024", "Sixes", 2, "Virat Kohli (RCB)", 38],
    ["2024", "Sixes", 3, "Heinrich Klaasen (SRH)", 38],
    ["2024", "Sixes", 4, "Nicholas Pooran (LSG)", 36],
    ["2024", "Sixes", 5, "Riyan Parag (RR)", 33],

    # IPL 2025 - Run scorers
    ["2025", "Runs", 1, "Sai Sudharsan (GT)", 759],
    ["2025", "Runs", 2, "Suryakumar Yadav (MI)", 717],
    ["2025", "Runs", 3, "Virat Kohli (RCB)", 657],
    ["2025", "Runs", 4, "Shubman Gill (GT)", 650],
    ["2025", "Runs", 5, "Mitchell Marsh (LSG)", 627],

    # IPL 2025 - Wicket takers
    ["2025", "Wickets", 1, "Prasidh Krishna (GT)", 25],
    ["2025", "Wickets", 2, "Noor Ahmad (CSK)", 24],
    ["2025", "Wickets", 3, "Josh Hazlewood (RCB)", 22],
    ["2025", "Wickets", 4, "Trent Boult (MI)", 22],
    ["2025", "Wickets", 5, "Arshdeep Singh (PBKS)", 21],

    # IPL 2025 - Six hitters
    ["2025", "Sixes", 1, "Nicholas Pooran (LSG)", 40],
    ["2025", "Sixes", 2, "Shreyas Iyer (PBKS)", 39],
    ["2025", "Sixes", 3, "Suryakumar Yadav (MI)", 38],
    ["2025", "Sixes", 4, "Mitchell Marsh (LSG)", 37],
    ["2025", "Sixes", 5, "Prabhsimran Singh (PBKS)", 30],

    # IPL 2026 - Run scorers
    ["2026", "Runs", 1, "Vaibhav Sooryavanshi (RR)", 776],
    ["2026", "Runs", 2, "Shubman Gill (GT)", 732],
    ["2026", "Runs", 3, "Sai Sudharsan (GT)", 722],
    ["2026", "Runs", 4, "Virat Kohli (RCB)", 675],
    ["2026", "Runs", 5, "Heinrich Klaasen (SRH)", 624],

    # IPL 2026 - Wicket takers
    ["2026", "Wickets", 1, "Kagiso Rabada (GT)", 29],
    ["2026", "Wickets", 2, "Bhuvneshwar Kumar (RCB)", 28],
    ["2026", "Wickets", 3, "Jofra Archer (RR)", 25],
    ["2026", "Wickets", 4, "Rashid Khan (GT)", 21],
    ["2026", "Wickets", 5, "Anshul Kamboj (CSK)", 21],

    # IPL 2026 - Six hitters
    ["2026", "Sixes", 1, "Vaibhav Sooryavanshi (RR)", 72],
    ["2026", "Sixes", 2, "Abhishek Sharma (SRH)", 43],
    ["2026", "Sixes", 3, "Rajat Patidar (RCB)", 42],
    ["2026", "Sixes", 4, "Ryan Rickelton (MI)", 38],
    ["2026", "Sixes", 5, "Mitchell Marsh (LSG)", 36]
]

player_df = pd.DataFrame(player_data, columns=[
    "Year", "Category", "Rank", "Player", "Value"
])


# -------------------- FUNCTIONS --------------------

# Function to display season summary
def show_season_summary():
    year = input("Enter year (2024, 2025 or 2026): ")
    filtered = season_df[season_df["Year"] == year]

    if filtered.empty:
        print("Invalid year.")
    else:
        row = filtered.iloc[0]
        print("\nIPL", year, "SEASON SUMMARY")
        print("Winner:", row["Winner"])
        print("Runner-up:", row["Runner-up"])
        print("Orange Cap:", row["Orange Cap"])
        print("Orange Cap Runs:", row["Orange Cap Runs"])
        print("Purple Cap:", row["Purple Cap"])
        print("Purple Cap Wickets:", row["Purple Cap Wickets"])
        print("Total Fours:", row["Fours"])
        print("Total Sixes:", row["Sixes"])
        print("Total Boundaries:", row["Total Boundaries"])


# Function to display playoff results
def show_playoff_results():
    year = input("Enter year (2024, 2025 or 2026): ")
    filtered = playoff_df[playoff_df["Year"] == year]

    print("\nPLAYOFF RESULTS OF", year)

    if filtered.empty:
        print("No data found.")
    else:
        for _, row in filtered.iterrows():
            print("\nMatch:", row["Match"])
            print("Scorecard:", row["Scorecard"])
            print("Venue:", row["Venue"])
            print("Winner:", row["Winner"])
            print("Result:", row["Result"])
            print("Player of the Match:", row["Player of the Match"])


# Function to display top five players
def show_top_players():
    year = input("Enter year (2024, 2025 or 2026): ")

    print("\nSelect category")
    print("1. Top run scorers")
    print("2. Top wicket takers")
    print("3. Top six hitters")

    choice = input("Enter choice: ")

    if choice == "1":
        category = "Runs"
    elif choice == "2":
        category = "Wickets"
    elif choice == "3":
        category = "Sixes"
    else:
        print("Wrong choice.")
        return

    filtered = player_df[(player_df["Year"] == year) & (player_df["Category"] == category)]

    print("\nTOP 5", category, "OF", year)

    if filtered.empty:
        print("No data found.")
    else:
        for _, row in filtered.iterrows():
            print(row["Rank"], row["Player"], "-", row["Value"])


# Function to compare boundaries
def compare_boundaries():
    print("\nBOUNDARY COMPARISON")

    years = season_df["Year"].to_numpy()
    fours = season_df["Fours"].to_numpy()
    sixes = season_df["Sixes"].to_numpy()
    totals = np.add(fours, sixes)

    for yr, tot in zip(years, totals):
        print(yr, ":", tot, "boundaries")

    max_idx = np.argmax(totals)
    highest = totals[max_idx]
    highest_year = years[max_idx]

    print("\nHighest boundaries:", highest)
    print("Season:", highest_year)


# -------------------- MAIN MENU --------------------

while True:
    print("\n================================")
    print("       IPL MATCH ANALYSIS")
    print("================================")
    print("1. View season summary")
    print("2. View playoff results")
    print("3. View top five players")
    print("4. Compare boundaries")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_season_summary()

    elif choice == "2":
        show_playoff_results()

    elif choice == "3":
        show_top_players()

    elif choice == "4":
        compare_boundaries()

    elif choice == "5":
        print("Thank you for using IPL Match Analysis.")
        break

    else:
        print("Invalid choice. Please try again.")
```
